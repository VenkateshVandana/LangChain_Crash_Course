# Agent

An agent is a model calling tools in a loop until a given task is complete.

```python
from langchain.agents import create_agent

agent_a= create_agent(
    model="groq:openai/gpt-oss-20b",
    tools=[]
    
)

res=agent_a.invoke({"messages":"What is Agentic AI?"})
res["messages"][-1].content
```

# Stream

```python
from langchain.agents import create_agent

agent_b= create_agent(
    model="groq:openai/gpt-oss-20b",
    tools=[]
    
)

res=agent_a.stream({"messages":"Write a essay on India?, explain 100 points"},stream_mode="messages")
for token,j in res:
    print(token.content, end="", flush=True)
```

# Batch Method

```python
from langchain.agents import create_agent

agent_b= create_agent(
    model="groq:openai/gpt-oss-20b",
    tools=[]
    
)

res=agent_a.batch([
    {"messages":"Write a essay on India?"},
    {"messages":"Write Humty Dumpty poem"}
    ])
for token in res:
    print(token["messages"][-1].content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

user_prompt = ChatPromptTemplate([
    ("system", "You are an expert in {Subject}."),
    ("human", "Provide information about {topic}."),
    ("ai", "Sure, I will provide details with accuracy.")
])

Subject = input("Enter Required Subject: ")
topic = input("Enter Required Topic: ")

prompt = user_prompt.invoke({
    "Subject": Subject,
    "topic": topic
})

res = agent_b.invoke({
    "messages": prompt.messages
})

print(res["messages"][-1].content)
```

```python
from langchain_core.messages import SystemMessage, HumanMessage
from langchain.agents import create_agent

agent_b = create_agent(
    model="groq:openai/gpt-oss-20b",
    tools=[]
)

messages = [
    SystemMessage(
        content="You are an expert Python teacher. Explain concepts simply."
    ),
    HumanMessage(
        content="What is a Python decorator?"
    )
]

res = agent_b.invoke({
    "messages": messages
})

print(res["messages"][-1].content)
```

# Tools

```python
from langchain.tools import tool
import requests

@tool
def get_weather(city:str) -> str:
    """Get Weather for a  City and according to the degree create ur own condition"""
    data = requests.get(f'https://wttr.in/{city}?format=j1').json()
    temp = data["current_condition"][0]["temp_C"]
    return f"{city}: {temp}c"

weather_agent= create_agent(
    model="groq:openai/gpt-oss-20b",
    tools=[get_weather],
    system_prompt="you are a weather agent and tells weather details of region"
)

res= weather_agent.invoke({"messages":"what is the weather in Palakonda"})
res
```

```python
print(res["messages"][-1].content, end="", flush=True)
```

# Structured Output

```python
from langchain.agents import create_agent
from langchain.tools import tool
from pydantic import BaseModel, Field
import requests


class WeatherResponse(BaseModel):
    city: str = Field(description="User provided city")
    temp: str = Field(description="Temperature of the city")
    cond: str = Field(
        description="Weather condition such as cold, warm, hot, humid"
    )
    suggestion: str = Field(
        description="Suggest suitable clothing and food for the climate"
    )


@tool
def get_weather(city: str) -> str:
    """Get the current temperature for a city."""

    data = requests.get(
        f"https://wttr.in/{city}?format=j1"
    ).json()

    temp = data["current_condition"][0]["temp_C"]

    return f"{city}: {temp}°C"


structured_weather_agent = create_agent(
    model="groq:qwen/qwen3.8-27b",
    tools=[get_weather],
    system_prompt="""
    You are a weather agent.

    When the user asks about weather:
    1. Use the get_weather tool.
    2. Determine the weather condition from the temperature.
    3. Suggest appropriate clothing and food.
    4. Return the final answer according to the WeatherResponse schema.
    """,
    response_format=WeatherResponse
)


res = structured_weather_agent.invoke({
    "messages": [
        ("user", "What is the weather in Palakonda?")
    ]
})

print(res["structured_response"])
```
